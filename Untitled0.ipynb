### Origen y preparación del dataset
Las imágenes se descargaron manualmente desde buscadores de internet usando la extensión de Chrome **Download All Images**, una carpeta por categoría (`arduino` y `raspberry_pi`). Luego se hizo una depuración con ayuda de ChatGPT para descartar las imágenes que no correspondían a la categoría. El resultado final fueron **339 imágenes** (150 de Arduino y 189 de Raspberry Pi), guardadas en Google Drive.

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

print("Versión de TensorFlow:", tf.__version__)
```

```python
# Monta Drive para acceder a las imágenes
from google.colab import drive
drive.mount('/content/drive')
```

```python
# ---------- PARÁMETROS (modificar solo esta celda) ----------
DATASET_PATH = '/content/drive/MyDrive/dataset_proyecto'  # carpeta con una subcarpeta por categoría
IMG_SIZE     = (128, 128)   # todas las imágenes se redimensionan a este tamaño
BATCH_SIZE   = 32           # imágenes procesadas por lote
EPOCHS       = 30           # máximo de épocas (EarlyStopping puede cortar antes)
VAL_SPLIT    = 0.2          # 20% de las imágenes se reserva para validación
SEED         = 42           # semilla para que el resultado sea reproducible
# ------------------------------------------------------------
```

```python
# Verificación rápida: cuántas imágenes hay por categoría
extensiones = ('.jpg', '.jpeg', '.png', '.bmp', '.gif')
for categoria in sorted(os.listdir(DATASET_PATH)):
    ruta = os.path.join(DATASET_PATH, categoria)
    if os.path.isdir(ruta):
        n = len([f for f in os.listdir(ruta) if f.lower().endswith(extensiones)])
        print(f"{categoria}: {n} imágenes")
```

```python
# Keras lee las carpetas y usa el nombre de cada una como etiqueta (clase)
train_ds = keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=VAL_SPLIT,
    subset='training',
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

val_ds = keras.utils.image_dataset_from_directory(
    DATASET_PATH,
    validation_split=VAL_SPLIT,
    subset='validation',
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)

class_names = train_ds.class_names
num_classes = len(class_names)
print("Categorías detectadas:", class_names)
```

```python
# Visualizamos algunas imágenes de ejemplo con su etiqueta
plt.figure(figsize=(10, 10))
for imagenes, etiquetas in train_ds.take(1):
    for i in range(min(9, len(imagenes))):
        plt.subplot(3, 3, i + 1)
        plt.imshow(imagenes[i].numpy().astype('uint8'))
        plt.title(class_names[etiquetas[i]])
        plt.axis('off')
plt.show()
```

```python
# Optimización de lectura: guarda en caché y prepara el siguiente lote mientras se entrena
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.cache().prefetch(buffer_size=AUTOTUNE)
val_ds   = val_ds.cache().prefetch(buffer_size=AUTOTUNE)
```

```python
# Aumento de datos (solo se aplica durante el entrenamiento)
data_augmentation = keras.Sequential([
    layers.RandomFlip('horizontal'),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
])

model = keras.Sequential([
    layers.Input(shape=IMG_SIZE + (3,)),
    data_augmentation,
    layers.Rescaling(1./255),

    layers.Conv2D(32, 3, activation='relu'),
    layers.MaxPooling2D(),

    layers.Conv2D(64, 3, activation='relu'),
    layers.MaxPooling2D(),

    layers.Conv2D(128, 3, activation='relu'),
    layers.MaxPooling2D(),

    layers.Dropout(0.3),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
])

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',  # etiquetas como números enteros (0, 1, 2...)
    metrics=['accuracy']
)
model.summary()
```

```python
# EarlyStopping: corta el entrenamiento si la validación deja de mejorar
early_stop = keras.callbacks.EarlyStopping(
    monitor='val_loss', patience=5, restore_best_weights=True
)

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=EPOCHS,
    callbacks=[early_stop]
)
```

```python
# Curvas de aprendizaje: sirven para detectar sobreajuste (overfitting)
acc, val_acc = history.history['accuracy'], history.history['val_accuracy']
loss, val_loss = history.history['loss'], history.history['val_loss']
epocas = range(1, len(acc) + 1)

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(epocas, acc, label='Entrenamiento')
plt.plot(epocas, val_acc, label='Validación')
plt.title('Precisión'); plt.xlabel('Época'); plt.legend()

plt.subplot(1, 2, 2)
plt.plot(epocas, loss, label='Entrenamiento')
plt.plot(epocas, val_loss, label='Validación')
plt.title('Pérdida'); plt.xlabel('Época'); plt.legend()
plt.show()
```

```python
# Recorremos la validación UNA sola vez guardando etiquetas reales, predicciones y errores
y_real, y_pred, errores = [], [], []

for imagenes, etiquetas in val_ds:
    probs = model.predict(imagenes, verbose=0)
    preds = np.argmax(probs, axis=1)
    for img, real, pred, p in zip(imagenes.numpy(), etiquetas.numpy(), preds, probs):
        y_real.append(real)
        y_pred.append(pred)
        if real != pred:
            errores.append((img.astype('uint8'), real, pred, p[pred]))

y_real, y_pred = np.array(y_real), np.array(y_pred)
exactitud = np.mean(y_real == y_pred)

print(f"Imágenes de validación: {len(y_real)}")
print(f"Aciertos: {np.sum(y_real == y_pred)}  |  Errores: {len(errores)}")
print(f"Exactitud (accuracy): {exactitud*100:.2f}%  |  Margen de error: {(1-exactitud)*100:.2f}%")
```

```python
# Matriz de confusión: filas = clase real, columnas = clase predicha
cm = confusion_matrix(y_real, y_pred)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predicción'); plt.ylabel('Real'); plt.title('Matriz de confusión')
plt.show()

# Precisión, recall y F1 por categoría
print(classification_report(y_real, y_pred, target_names=class_names, zero_division=0))
```

```python
# Imágenes mal clasificadas (útil para entender POR QUÉ falla el modelo)
if errores:
    n = min(12, len(errores))
    plt.figure(figsize=(14, 3.5 * ((n + 3) // 4)))
    for i, (img, real, pred, conf) in enumerate(errores[:n]):
        plt.subplot((n + 3) // 4, 4, i + 1)
        plt.imshow(img)
        plt.title(f"Real: {class_names[real]}\nPredicho: {class_names[pred]} ({conf*100:.0f}%)", fontsize=9)
        plt.axis('off')
    plt.tight_layout()
    plt.show()
else:
    print("El modelo no cometió errores en validación.")
```
